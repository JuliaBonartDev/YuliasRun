# 🤔 ¿Por qué POO?

Imagina que tienes que programar un juego. **Sin POO**, tendrías algo así:

````python
# ❌ Código sin POO - ¡Un desastre!
jugador_x = 100
jugador_y = 200
jugador_vidas = 3
jugador_puntos = 0

enemigo1_x = 50
enemigo1_y = 50
enemigo1_vidas = 2

enemigo2_x = 150
enemigo2_y = 75
enemigo2_vidas = 2

def mover_jugador(direccion):
    global jugador_x, jugador_y
    if direccion == "izquierda":
        jugador_x -= 10
    # ... más código repetitivo

def mover_enemigo1(direccion):
    global enemigo1_x, enemigo1_y
    # ... código casi idéntico al anterior

def mover_enemigo2(direccion):
    global enemigo2_x, enemigo2_y
    # ... ¡más código repetido!

print("😱 Imagina tener 20 enemigos... ¡Sería un caos!")
````

## ✨ La POO nos salva porque permite:

### 🗂️ **Organizar código**
- Cada "cosa" (jugador, enemigo) tiene su lugar

### 🏗️ **Modelar "cosas" del mundo real**
- Un jugador **es** algo que tiene posición, vidas, y puede moverse

### ♻️ **Reutilizar código**
- Una vez que defines "Enemigo", puedes crear 100 sin repetir código

### 🔧 **Mantener y cambiar fácilmente**
- Si cambias cómo se mueve un enemigo, ¡cambia para todos automáticamente!

# 🏗️ Tu Primera Clase: Jugador

¡Hora de crear tu primera clase! Una **clase** es como un **molde para galletas** 🍪 - defines la forma una vez, y puedes hacer todas las galletas que quieras.

## 🎭 De problema real a solución POO

**Problema**: Necesitamos muchos personajes en nuestro juego  
**Solución**: ¡Crear una clase `Jugador`!

````python
# ✨ ¡Nuestra primera clase!
class Jugador:
    def __init__(self, nombre):
        """
        Constructor: se ejecuta cuando creamos un nuevo jugador
        """
        print(f"🎮 ¡Creando jugador {nombre}!")
        
        # 📦 ATRIBUTOS: características del jugador
        self.nombre = nombre
        self.vidas = 3
        self.puntos = 0
        self.x = 100  # posición horizontal
        self.y = 200  # posición vertical
        
        print(f"✅ {self.nombre} está listo con {self.vidas} vidas")
    
    # ⚡ MÉTODOS: acciones que puede hacer el jugador
    def mover(self, direccion):
        """Mover el jugador en una dirección"""
        if direccion == "izquierda":
            self.x -= 10
            print(f"👈 {self.nombre} se movió a la izquierda (x={self.x})")
        elif direccion == "derecha":
            self.x += 10
            print(f"👉 {self.nombre} se movió a la derecha (x={self.x})")
        elif direccion == "arriba":
            self.y -= 10
            print(f"👆 {self.nombre} subió (y={self.y})")
        elif direccion == "abajo":
            self.y += 10
            print(f"👇 {self.nombre} bajó (y={self.y})")
    
    def puntuar(self, puntos):
        """Añadir puntos al jugador"""
        self.puntos += puntos
        print(f"🌟 ¡{self.nombre} ganó {puntos} puntos! Total: {self.puntos}")
    
    def herir(self):
        """Quitar una vida al jugador"""
        if self.vidas > 0:
            self.vidas -= 1
            print(f"💔 ¡{self.nombre} perdió una vida! Quedan: {self.vidas}")
            if self.vidas == 0:
                print(f"💀 ¡{self.nombre} ha perdido el juego!")
        return self.vidas > 0  # True si sigue vivo

print("🎯 ¡Clase Jugador creada! Ahora vamos a usarla...")
````

## 🎪 ¡Hora del espectáculo! Creemos personajes

Ejecuta la celda de arriba y luego esta:

````python
# 🎭 Creando OBJETOS desde la clase (como galletas del molde)
print("=== 🏭 FÁBRICA DE JUGADORES ===")

# Crear dos jugadores diferentes
julia = Jugador("Julia")
cale = Jugador("Cale")

print("\n=== 🎮 ESTADO INICIAL ===")
print(f"📊 {julia.nombre}: {julia.vidas} vidas, {julia.puntos} puntos, posición ({julia.x}, {julia.y})")
print(f"📊 {cale.nombre}: {cale.vidas} vidas, {cale.puntos} puntos, posición ({cale.x}, {cale.y})")

print("\n💡 ¡Fíjate! Son dos OBJETOS diferentes con el mismo 'molde' (clase)")
print(f"🔍 julia es de tipo: {type(julia)}")
print(f"🔍 cale es de tipo: {type(cale)}")
````

# 🕵️ El Misterio de `self` - ¡Resuelto!

**Pregunta del millón**: ¿Por qué todos los métodos empiezan con `self`? 🤔

## 🎭 Imaginemos una conversación...

**Roxi**: "¡Oye, muévete a la derecha!"  
**¿Quién se mueve?** ¿Roxi? ¿Marcus? ¿Rubén? 

**En POO es igual**: Cuando llamas `mover("derecha")`, ¿quién se mueve?

`self` = **"YO, el objeto que está ejecutando este método"**

````python
# 🔍 ¡Experimento para entender self!
print("=== 🧪 EXPERIMENTO SELF ===")

# Creamos varios estudiantes
miguel = Jugador("Miguel")
marcus = Jugador("Marcus")
dani = Jugador("Dani")

print("\n=== 🎯 POSICIONES INICIALES ===")
print(f"Miguel está en x={miguel.x}")
print(f"Marcus está en x={marcus.x}")  
print(f"Dani está en x={dani.x}")

print("\n=== 🏃 ¡CADA UNO SE MUEVE INDEPENDIENTEMENTE! ===")

# Cada uno se mueve - ¡observa los mensajes!
miguel.mover("derecha")  # self será miguel
marcus.mover("izquierda")   # self será marcus  
dani.mover("derecha")    # self será dani

print("\n=== 📍 POSICIONES FINALES ===")
print(f"Miguel terminó en x={miguel.x}")
print(f"Marcus terminó en x={marcus.x}")
print(f"Dani terminó en x={dani.x}")

print("\n💡 ¡Fíjate! Cada objeto modificó SU PROPIO estado, no el de los otros")
print("🎯 ESO es lo que hace 'self': indica 'mi propio estado'")
````

## 🎮 Mini-Ejercicio: ¡Añade velocidad!

**Tu misión**: Mejora la clase `Jugador` añadiendo:
- Un atributo `velocidad` (por defecto 10)  
- Un método `acelerar(factor)` que multiplique la velocidad
- Modifica `mover()` para usar la velocidad actual

**¡Piensa!** ¿Dónde va cada cosa?

````python
# 🚀 ¡Mejoremos la clase Jugador!
class JugadorMejorado:
    def __init__(self, nombre):
        print(f"🎮 ¡Creando jugador mejorado {nombre}!")
        
        # Atributos existentes
        self.nombre = nombre
        self.vidas = 3
        self.puntos = 0
        self.x = 100
        self.y = 200
        
        # 🆕 ¡NUEVO! Atributo velocidad
        self.velocidad = 10  # píxeles por movimiento
        
        print(f"✅ {self.nombre} está listo con velocidad {self.velocidad}")
    
    def mover(self, direccion):
        """Mover usando la velocidad actual"""
        if direccion == "izquierda":
            self.x -= self.velocidad  # ¡Usa self.velocidad!
            print(f"👈 {self.nombre} se movió {self.velocidad} píxeles a la izquierda (x={self.x})")
        elif direccion == "derecha":
            self.x += self.velocidad
            print(f"👉 {self.nombre} se movió {self.velocidad} píxeles a la derecha (x={self.x})")
        # ... (arriba y abajo igual)
    
    def acelerar(self, factor):
        """🚀 Acelerar el jugador"""
        self.velocidad *= factor
        print(f"🏃‍♀️ ¡{self.nombre} ahora va a velocidad {self.velocidad}!")
    
    def puntuar(self, puntos):
        self.puntos += puntos
        print(f"🌟 ¡{self.nombre} ganó {puntos} puntos! Total: {self.puntos}")
    
    def herir(self):
        if self.vidas > 0:
            self.vidas -= 1
            print(f"💔 ¡{self.nombre} perdió una vida! Quedan: {self.vidas}")
        return self.vidas > 0

print("🎯 ¡Clase mejorada creada!")
````

````python
# 🏁 ¡Carrera de velocidad entre compañeros!
print("=== 🏁 CARRERA DE VELOCIDAD ===")

# Creamos corredores de la clase
presi = JugadorMejorado("Presi")
ramazan = JugadorMejorado("Ramazan") 
wizi = JugadorMejorado("Wizi")

print("\n=== 🚀 ¡MODIFICAMOS VELOCIDADES! ===")
presi.acelerar(2)        # Presi va 2x más rápido
ramazan.acelerar(1.5)    # Ramazan 1.5x más rápido  
# Wizi mantiene velocidad normal

print("\n=== 🏃 ¡A CORRER! ===")
print("🏁 Cada uno da 3 pasos a la derecha...")

for paso in range(1, 4):
    print(f"\n--- Paso {paso} ---")
    presi.mover("derecha")
    ramazan.mover("derecha")
    wizi.mover("derecha")

print("\n=== 🏆 RESULTADOS FINALES ===")
print(f"🥇 Presi: posición x={presi.x} (velocidad {presi.velocidad})")
print(f"🥈 Ramazan: posición x={ramazan.x} (velocidad {ramazan.velocidad})")
print(f"🥉 Wizi: posición x={wizi.x} (velocidad {wizi.velocidad})")

print("\n💡 ¡Cada objeto mantiene SU PROPIA velocidad!")
print("🎯 Esto es encapsulación: cada objeto controla su propio estado")
````

# 🔫 Composición Simple: ¡Objetos que usan otros objetos!

**Nuevo reto**: Nuestros jugadores necesitan **cuchillos** para defenderse.

¿Cómo modelamos esto? 🤔

## 🧩 Solución: ¡Una clase usa otra clase!

- **Jugador** → puede crear y usar **Cuchillos**
- **Cuchillo** → tiene su propia lógica (dirección, velocidad)

**Esto es COMPOSICIÓN**: Un objeto "tiene" otros objetos

````python
# 🔪 ¡Creemos la clase Cuchillo!
class Cuchillo:
    def __init__(self, x, y, direccion):
        """Un cuchillo tiene posición y dirección"""
        self.x = x
        self.y = y
        self.direccion = direccion
        self.velocidad = 15  # Los cuchillos van rápido
        self.activo = True   # ¿Sigue volando?
        
        print(f"🔪 ¡Cuchillo lanzado desde ({x}, {y}) hacia {direccion}!")
    
    def update(self):
        """Mover el cuchillo según su dirección"""
        if self.activo:
            if self.direccion == "arriba":
                self.y -= self.velocidad
            elif self.direccion == "abajo":  
                self.y += self.velocidad
            elif self.direccion == "izquierda":
                self.x -= self.velocidad
            elif self.direccion == "derecha":
                self.x += self.velocidad
            
            # ¿Se salió de la pantalla?
            if self.y < 0 or self.y > 400 or self.x < 0 or self.x > 800:
                self.activo = False
                print(f"💨 Cuchillo salió de pantalla en ({self.x}, {self.y})")

# 🎮 ¡Jugador que puede lanzar cuchillos!
class JugadorConCuchillos:
    def __init__(self, nombre):
        self.nombre = nombre
        self.x = 100
        self.y = 200
        self.cuchillos = []  # 📦 Lista de cuchillos lanzados
        self.enfriamiento = 0  # Contador para evitar spam
        
        print(f"🎮 {self.nombre} está listo para lanzar cuchillos!")
    
    def lanzar_cuchillo(self, direccion):
        """🔫 Lanzar un cuchillo si no hay enfriamiento"""
        if self.enfriamiento <= 0:
            # Crear un nuevo cuchillo en mi posición
            nuevo_cuchillo = Cuchillo(self.x, self.y, direccion)
            self.cuchillos.append(nuevo_cuchillo)
            self.enfriamiento = 3  # 3 turnos de enfriamiento
            print(f"🎯 {self.nombre} lanzó un cuchillo hacia {direccion}")
        else:
            print(f"⏰ {self.nombre} debe esperar {self.enfriamiento} turnos")
    
    def update(self):
        """Actualizar enfriamiento y todos los cuchillos"""
        # Reducir enfriamiento
        if self.enfriamiento > 0:
            self.enfriamiento -= 1
        
        # Actualizar todos los cuchillos
        for cuchillo in self.cuchillos:
            cuchillo.update()
        
        # Limpiar cuchillos inactivos
        cuchillos_activos = [c for c in self.cuchillos if c.activo]
        eliminados = len(self.cuchillos) - len(cuchillos_activos)
        self.cuchillos = cuchillos_activos
        
        if eliminados > 0:
            print(f"🧹 {self.nombre} limpió {eliminados} cuchillos inactivos")

print("🎯 ¡Clases de combate creadas!")
````

````python
# ⚔️ ¡Duelo de cuchillos entre Rubén y David!
print("=== ⚔️ DUELO DE CUCHILLOS ===")

ruben = JugadorConCuchillos("Rubén")
david = JugadorConCuchillos("David")

print("\n=== 🎯 RONDA 1: LANZAMIENTOS ===")
ruben.lanzar_cuchillo("arriba")
david.lanzar_cuchillo("derecha")
ruben.lanzar_cuchillo("izquierda")  # ¿Podrá lanzar tan rápido?

print(f"\n📊 Rubén tiene {len(ruben.cuchillos)} cuchillos volando")
print(f"📊 David tiene {len(david.cuchillos)} cuchillos volando")

print("\n=== ⏰ SIMULANDO 5 TURNOS ===")
for turno in range(1, 6):
    print(f"\n--- Turno {turno} ---")
    
    # Actualizar ambos jugadores
    ruben.update()
    david.update()
    
    # Intentar lanzar más cuchillos
    if turno == 2:
        david.lanzar_cuchillo("arriba")
    if turno == 4:
        ruben.lanzar_cuchillo("derecha")
        david.lanzar_cuchillo("izquierda")
    
    print(f"📊 Estado turno {turno}:")
    print(f"   Rubén: {len(ruben.cuchillos)} cuchillos, enfriamiento: {ruben.enfriamiento}")
    print(f"   David: {len(david.cuchillos)} cuchillos, enfriamiento: {david.enfriamiento}")

print("\n💡 ¡Fíjate los conceptos de POO!")
print("🔸 COMPOSICIÓN: Jugador 'tiene' una lista de Cuchillos")
print("🔸 RESPONSABILIDADES: Cada clase sabe qué hacer")
print("🔸 ESTADO: Cada objeto mantiene sus propios datos")
````

# 📁 Modularidad: ¡Orden en el código!

**Problema**: Si seguimos añadiendo clases, ¡el archivo se hará gigantesco! 😱

**Solución**: Separar en archivos según su propósito

## 🗂️ Organización típica en un juego:

```
mi_juego/
├── entities.py     # 👾 Jugador, Obstáculo, Cuchillo, PowerUp
├── abilities.py    # ⚡ Efectos, Cooldowns, Combos
├── game_states.py  # 🎯 Menú, Jugando, Pausa, GameOver
├── settings.py     # ⚙️ Constantes, configuración
└── main.py         # 🎮 Game loop principal
```

### 💡 ¿Por qué separar?

- **🔍 Fácil de encontrar**: "Las clases de personajes están en entities.py"
- **👥 Trabajo en equipo**: Cada persona puede trabajar en un archivo
- **🔧 Mantenimiento**: Cambiar un power-up no toca el código del jugador
- **♻️ Reutilización**: Puedes usar las clases en otros proyectos
